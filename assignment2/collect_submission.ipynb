```python
# Environment setup (Kaggle / local).
import os
import sys
from pathlib import Path


def find_assignment_root():
    candidates = [
        Path.cwd(),
        Path.cwd() / "assignment2",
        Path("/kaggle/working"),
        Path("/kaggle/working/CS231n/assignment2"),
    ]
    for candidate in candidates:
        if (candidate / "cs231n").exists():
            return candidate
    for candidate in Path("/kaggle/working").glob("**/assignment2"):
        if (candidate / "cs231n").exists():
            return candidate
    return Path.cwd()


ROOT = find_assignment_root()
os.chdir(ROOT)
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))
print("Working directory:", Path.cwd())
```
